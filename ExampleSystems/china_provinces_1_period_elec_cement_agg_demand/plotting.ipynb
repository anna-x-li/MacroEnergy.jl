```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import plotly.express as px

df = cement_flows
provinces = [
    'Beijing', 'Tianjin', 'Hebei', 'Shanxi', 'Inner Mongolia', 'Liaoning', 'Jilin', 'Heilongjiang',
    'Shanghai', 'Jiangsu', 'Zhejiang', 'Anhui', 'Fujian', 'Jiangxi', 'Shandong', 'Henan',
    'Hubei', 'Hunan', 'Guangdong', 'Guangxi', 'Hainan', 'Chongqing', 'Sichuan', 'Guizhou',
    'Yunnan', 'Shaanxi', 'Gansu', 'Qinghai', 'Ningxia', 'Xinjiang'
]

# --- Select columns to plot ---
cols = df.columns[:32]  # plot first 32 columns
nrows, ncols = 4, 8

# --- Color palette ---
colors = px.colors.qualitative.Plotly * ((len(cols) // len(px.colors.qualitative.Plotly)) + 1)

# --- Create 8×32 subplot grid with added padding ---
fig = make_subplots(
    rows=nrows, cols=ncols,
    subplot_titles=provinces,
    vertical_spacing=0.07,   # increase vertical spacing (padding)
    horizontal_spacing=0.05  # increase horizontal spacing (padding)
)

# --- Add traces ---
for i, col in enumerate(cols):
    row = i % nrows + 1
    col_pos = i // nrows + 1
    color = colors[i]
    fig.add_trace(
        go.Scatter(
            x=df.index,
            y=df[col],
            mode='lines',
            name=col,
            line=dict(width=1.2, color=color)
        ),
        row=row, col=col_pos
    )

# --- Layout & style tweaks ---
fig.update_layout(
    height=900,
    width=1400,  # keep it compact but readable
    showlegend=False,
    title_text="Cement Flow by Province (8×32 grid, with padding)",
    plot_bgcolor='white',
    paper_bgcolor='white',
)

# --- Clean look: white background, black frame, no gridlines ---
fig.update_xaxes(showgrid=False, zeroline=False, showline=True, linecolor='black', mirror=True)
fig.update_yaxes(showgrid=False, zeroline=False, showline=True, linecolor='black', mirror=True, matches='y')

fig.show()

```

```python
def plot_lines(dfs, provinces, title="", labels=None):
    nrows, ncols = 8, 4
    n_subplots = 31
    
    # --- Default labels if not provided ---
    if labels is None:
        labels = [f"Scenario {i+1}" for i in range(len(dfs))]
    
    # --- Colors for each dataset ---
    line_colors = px.colors.qualitative.Plotly[:len(dfs)]
    
    # --- Create subplot grid ---
    fig = make_subplots(
        rows=nrows, cols=ncols,
        subplot_titles=provinces,
        vertical_spacing=0.05,
        horizontal_spacing=0.05
    )
    
    # --- Add traces ---
    for i in range(n_subplots):
        row = i // ncols + 1
        col_pos = i % ncols + 1
        
        for j, df in enumerate(dfs):
            if i >= len(df.columns):
                continue  # skip if df has fewer columns
            
            fig.add_trace(
                go.Scatter(
                    x=df.index,
                    y=df.iloc[:, i],
                    mode='lines',
                    line=dict(width=1.5, color=line_colors[j]),
                    name=labels[j],
                    showlegend=(i == 0)  # show legend only once
                ),
                row=row, col=col_pos
            )
    
    # --- Layout and style ---
    fig.update_layout(
        height=2400, width=1600,
        showlegend=True,
        title_text=title,
        plot_bgcolor='white',
        paper_bgcolor='white',
        legend=dict(
            orientation="h",
            yanchor="bottom",
            y=-0.02,
            xanchor="center",
            x=0.5
        )
    )
    
    # --- Axis formatting ---
    fig.update_xaxes(showgrid=False, zeroline=False, showline=True, linecolor='black', mirror=True)
    fig.update_yaxes(showgrid=False, zeroline=False, showline=True, linecolor='black', mirror=True)
    
    fig.show()
```

```python
dfs = [pv_availability, pv_distribution_availability, wind_onshore_availability, wind_offshore, hydro_avilability]

plot_lines(dfs, provinces, title="VRE Availability by Province", labels=['pv', 'pv_distributed', 'wind_onshore', 'wind_offshore', 'hydro'])
```
