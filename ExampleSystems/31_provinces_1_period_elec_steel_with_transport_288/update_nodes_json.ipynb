```python
def load_json(filepath):
    with open(filepath, 'r') as f:
        return json.load(f)

def save_json(data, filepath):
    with open(filepath, 'w') as f:
        json.dump(data, f, indent=4, ensure_ascii=False)
```

```python
def add_crudesteel_instances(data, demand_dict, province_mapping):
    """Add instance_data for CrudeSteel demand for each province."""
    for node in data['nodes']:
        if node.get('type') == 'CrudeSteel':
            break
    else:
        print('No CrudeSteel node found!')
        return

    for province, region_id in province_mapping.items():
        if province in demand_dict:
            instance = {
                "constraints": {"AggregatedDemandConstraint": True},
                "rhs_policy": {"AggregatedDemandConstraint": int(demand_dict[province] * 288 / 8760)},
                "id": f"crudesteel_{region_id}"
            }
            node['instance_data'].append(instance)
            print(f"Added CrudeSteel demand for {province} ({region_id}): {demand_dict[province]}")

def add_steelscrap_instances(data, supply_dict, province_mapping):
    """Add instance_data for SteelScrap max_supply for each province."""
    for node in data['nodes']:
        if node.get('type') == 'SteelScrap':
            break
    else:
        print('No SteelScrap node found!')
        return

    for province, region_id in province_mapping.items():
        if province in supply_dict:
            instance = {
                "id": f"steelscrap_source_{region_id}",
                # "max_supply": [int(supply_dict[province])],
                # "price_supply": [0.00023593]
            }
            node['instance_data'].append(instance)
            print(f"Added SteelScrap supply for {province} ({region_id}): {supply_dict[province]}")
```

```python
# Example usage:
json_path = "system/nodes_1_template.json"
data = load_json(json_path)

# Add CrudeSteel demand and SteelScrap supply
add_crudesteel_instances(data, crude_steel_demand, province_mapping)
add_steelscrap_instances(data, steel_scrap_supply, province_mapping)

# Save to a new file to avoid overwriting original
save_json(data, json_path.replace('.json', '_updated.json'))
print(f"\nUpdated file saved as {json_path.replace('.json', '_updated.json')}")
```
