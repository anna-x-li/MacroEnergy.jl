```python
def create_instance_template(province, region_id, capacity, asset_type):
    """Create an instance template for a province
    
    Args:
        province: Province name
        region_id: Region identifier (e.g., "Region15Shandong")
        capacity: Existing capacity in tons/hour
        asset_type: Type of asset - "bfbof", "bfbofccs", "scrap_eaf_56", or "scrap_eaf_100"
    """
    
    # Define edge configurations by asset type
    edges_config = {
        "bfbof": {
            "crudesteel_edge": {
                "end_vertex": f"crudesteel_{region_id}",
                "existing_capacity": capacity
            },
            "metcoal_edge": {"start_vertex": f"coal_{region_id}"},
            "thermalcoal_edge": {"start_vertex": f"coal_{region_id}"},
            "elec_edge": {"start_vertex": f"elec_{region_id}"},
            "natgas_edge": {"start_vertex": f"natgas_{region_id}"},
            "ironore_edge": {"start_vertex": "ironore_source"},
            "steelscrap_edge": {"start_vertex": f"steelscrap_source_{region_id}"},
            "retrofit_options": [
                {
                    "template_id": f"bfbofccs_{region_id}",
                    "id": f"bfbofccs_retrofit_{region_id}",
                    "edges": {
                        "crudesteel_edge": {
                            "investment_cost": 3278693,
                            "is_retrofit": True,
                            "retrofit_efficiency": 1,
                            "can_expand": True
                        }
                    }
                }
            ]
        },
        "bfbofccs": {
            "crudesteel_edge": {
                "end_vertex": f"crudesteel_{region_id}",
                "existing_capacity": capacity
            },
            "metcoal_edge": {"start_vertex": f"coal_{region_id}"},
            "thermalcoal_edge": {"start_vertex": f"coal_{region_id}"},
            "elec_edge": {"start_vertex": f"elec_{region_id}"},
            "natgas_edge": {"start_vertex": f"natgas_{region_id}"},
            "ironore_edge": {"start_vertex": "ironore_source"},
            "steelscrap_edge": {"start_vertex": f"steelscrap_source_{region_id}"},
        },
        "scrap_eaf_56": {
            "crudesteel_edge": {
                "end_vertex": f"crudesteel_{region_id}",
                "existing_capacity": capacity
            },
            "steelscrap_edge": {"start_vertex": f"steelscrap_source_{region_id}"},
            "elec_edge": {"start_vertex": f"elec_{region_id}"},
            "naturalgas_edge": {"start_vertex": f"natgas_{region_id}"},
            "carbonsource_edge": {"start_vertex": f"coal_{region_id}"},
            "retrofit_options" : [
                {
                    "template_id": f"scrapeaf_100_{region_id}",
                    "id": f"scrapeaf_100_retrofit_{region_id}",
                    "edges": {
                        "crudesteel_edge": {
                            "is_retrofit": True,
                            "retrofit_efficiency": 1,
                            "can_expand": True
                        }
                    }
                }
            ]
        },
        "scrap_eaf_100": {
            "crudesteel_edge": {
                "end_vertex": f"crudesteel_{region_id}",
                "existing_capacity": capacity
            },
            "steelscrap_edge": {"start_vertex": f"steelscrap_source_{region_id}"},
            "elec_edge": {"start_vertex": f"elec_{region_id}"},
            "naturalgas_edge": {"start_vertex": f"natgas_{region_id}"},
            "carbonsource_edge": {"start_vertex": f"coal_{region_id}"},
        }
    }
    
    instance = {
        "id": f"{asset_type}_{region_id}",
        "edges": edges_config.get(asset_type, edges_config["bfbof"])
    }
    return instance
```

```python
def load_json(filepath):
    """Load the JSON file"""
    with open(filepath, 'r') as f:
        return json.load(f)

def save_json(data, filepath):
    """Save the JSON file"""
    with open(filepath, 'w') as f:
        json.dump(data, f, indent=4)
```

```python
def add_province_instance(data, province, region_id, capacity, asset_type):
    """Add a single province instance to the JSON"""
    json_key = asset_to_json_key.get(asset_type, asset_to_json_key[asset_type])
    instance = create_instance_template(province, region_id, capacity, asset_type)
    
    if json_key in data and len(data[json_key]) > 0:
        data[json_key][0]["instance_data"].append(instance)
        print(f"✓ Added instance for {province} ({region_id}) with capacity {capacity} tons/hour")
    else:
        print(f"✗ Error: Could not find key '{json_key}' in JSON data")

def update_all_provinces(filepath, asset_type, capacity_dict):
    """Update the JSON file with all provinces"""
    data = load_json(filepath)
    added_provinces = []
    
    for province, region_id in province_mapping.items():
        if province in capacity_dict:
            capacity = capacity_dict[province]
            add_province_instance(data, province, region_id, capacity, asset_type)
            added_provinces.append(province)
        else:
            print(f"⚠ Skipped {province} - no capacity data available")
    
    save_json(data, filepath.split(".")[0] + "_updated.json")
    print(f"\n✓ Successfully added {len(added_provinces)} provinces")
    print(f"File saved to: {filepath.split('.')[0]}_updated.json")
    
    return len(added_provinces)

def verify_instances(filepath, asset_type="bfbof"):
    """Verify the instances in the JSON file"""
    json_key = asset_to_json_key.get(asset_type, asset_to_json_key[asset_type])
    data = load_json(filepath)
    
    if json_key not in data or len(data[json_key]) == 0:
        print(f"✗ Error: Could not find key '{json_key}' in JSON data")
        return
    
    instances = data[json_key][0]["instance_data"]
    
    print(f"\nTotal instances: {len(instances)}\n")
    print("Instance Summary:")
    print("-" * 80)
    for instance in instances:
        instance_id = instance["id"]
        capacity = instance["edges"]["crudesteel_edge"].get("existing_capacity", "N/A")
        print(f"  {instance_id}: capacity = {capacity}")
```
